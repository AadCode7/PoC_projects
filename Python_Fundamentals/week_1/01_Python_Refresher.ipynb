This notebook covers the foundational Python concepts essential for data science.

## 1. Functions
Understanding how to define and use functions is essential.

```python
# Example: Basic Function
def greet(name):
    return f"Hello, {name}!"

greet("Alice")
```

## 2. List Comprehensions
Compact way to create lists using for loops.

```python
# Traditional for loop
squares = []
for i in range(10):
    squares.append(i ** 2)

# List comprehension
squares_lc = [i ** 2 for i in range(10)]

print(squares)
print(squares_lc)
```

## 3. Lambda Functions
Anonymous functions used for short snippets.

```python
# Lambda function
add = lambda x, y: x + y
add(5, 3)
```

## 4. Map, Filter, Reduce
Functional tools for transforming data.

```python
from functools import reduce

nums = [1, 2, 3, 4, 5]

# Map: square each number
squared = list(map(lambda x: x**2, nums))

# Filter: keep only even numbers
evens = list(filter(lambda x: x % 2 == 0, nums))

# Reduce: compute product of all numbers
product = reduce(lambda x, y: x * y, nums)

squared, evens, product
```

## 5. Decorators
Functions that modify the behavior of other functions.

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with args: {args}")
        return func(*args, **kwargs)
    return wrapper

@logger
def multiply(x, y):
    return x * y

multiply(3, 4)
```

## 6. Classes and Objects
Basics of Object-Oriented Programming (OOP) in Python.

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} says woof!"

dog1 = Dog("Buddy")
dog1.speak()
```
