# 🧠 Object-Oriented Programming (OOP) in Python

This notebook covers key OOP concepts in Python with examples:

- Class & Object
- Method
- Inheritance
- Polymorphism
- Abstraction
- Encapsulation

## 🧱 Class & Object

A **class** is a blueprint for creating objects.
An **object** is an instance of a class.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"

# Creating object
dog = Animal("Dog")
print(dog.speak())
```

## ⚙️ Methods

Functions defined in a class are called **methods**.
`__init__()` is a special method called the constructor.

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b

calc = Calculator()
print("Add:", calc.add(2, 3))
print("Multiply:", calc.multiply(2, 3))
```

## 🧬 Inheritance

**Inheritance** lets one class inherit properties/methods from another.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak1(self):
        return f"{self.name} speaks"

class Dog(Animal):
    def speak(self):
        return f"{self.name} barks"

dog = Dog("Buddy")
print(dog.speak1())
```

## 🔁 Polymorphism

**Polymorphism** means same method name behaves differently for different classes.

```python
class Cat:
    def speak(self):
        return "Meow"

class Dog:
    def speak(self):
        return "Bark"

animals = [Cat(), Dog()]
for animal in animals:
    print(animal.speak())
```

## 🎭 Abstraction

**Abstraction** hides complex implementation using abstract base classes. They act as placeholders that force subclasses to provide their own specific implementation, ensuring consistent structure across derived classes.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        return "Car started"

my_car = Car()
print(my_car.start())
```

## 🔒 Encapsulation

**Encapsulation** restricts access to methods and variables using `_` and `__`. Encapsulation means hiding internal details of a class and only exposing what’s necessary. Protects data from unauthorized access and accidental modification.

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount()
account.deposit(100)
print(account.get_balance())
```
