```python
# 📦 Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import glob
```

## 🔁 MultiIndexing and Hierarchical Indexing

```python
# Create a sample dataset
arrays = [
    ['Math', 'Math', 'Science', 'Science'],
    ['Test1', 'Test2', 'Test1', 'Test2']
]
index = pd.MultiIndex.from_arrays(arrays, names=('Subject', 'Test'))
df = pd.DataFrame({'Score': [88, 92, 85, 90]}, index=index)
df
```

### 🧠 Interview Question:
**Q: How would you access all `Test1` scores?**
**A:** `df.xs('Test1', level='Test')`

## 📈 Window Functions: Rolling & Expanding

```python
# Create a simple time-series-like data
np.random.seed(42)
data = pd.Series(np.random.randint(20, 100, size=10))
data.rolling(window=3).mean()
```

**Expanding window:** Useful for cumulative statistics

```python
data.expanding().mean()
```

## ⏳ Time Series with Pandas

```python
# Create a time-indexed DataFrame
date_rng = pd.date_range(start='2023-01-01', end='2023-01-10', freq='D')
ts_df = pd.DataFrame(date_rng, columns=['date'])
ts_df['value'] = np.random.randint(10, 100, size=(len(date_rng)))
ts_df = ts_df.set_index('date')
ts_df.head()
```

### 📦 Resampling
`resample('2D')` means 2-day frequency

```python
ts_df.resample('2D').mean()
```

### Interview Q: Shift vs Diff
**Q: What does `shift(1)` do?**
**A:** Moves data forward by 1 time unit.

**Q: What does `diff(1)` do?**
**A:** Computes the difference between current and previous values.

## 🚀 Interview-Level Exercises + Answers

```python
# Q1. Compute 7-day rolling average of the `value` column.
ts_df['7D_avg'] = ts_df['value'].rolling(window=7).mean()
ts_df.head(10)
```

```python
# Q2. Create a MultiIndexed DataFrame from a CSV-style data.
sales = pd.DataFrame({
    'Region': ['East', 'East', 'West', 'West'],
    'Product': ['A', 'B', 'A', 'B'],
    'Sales': [200, 300, 400, 250]
}).set_index(['Region', 'Product'])
sales
```

**Q: How do you retrieve all sales for Product A?**
**A:** `sales.xs('A', level='Product')`
