### 1] 2 SUM

```python
nums = [15,7,2,11]
target = 9

map1 = {}

def two_sum(nums,target):
    for i, num in enumerate(nums):
        if target - num in map1:
            return [map1[target-num],i]
        map1[num] = i
        print("Current Map:",map1)
        
print(two_sum(nums,target))
```

### 2] Contains Duplicate I

```python
nums = [1,4,3,2,7,5,5,7]

def find_duplicates(nums):
    seen = set()
    duplicates = set()

    for num in nums:
        if num in seen:
            duplicates.add(num)
            print(f"Duplicate Set found: {num}")
        else:
            seen.add(num)
        print(f"Current Map: {seen}")

find_duplicates(nums)

# Can also use Hashmap
```

### 3] Contains Duplicate II

Use the HashMap "Del" function

```python
# nums[i] == nums[j] and abs(i - j) <= k

nums = [1,2,3,4,5,5]
k = 3

# Using Sliding Window Approach
def contains_nearby_duplicate(nums, k):
    map3 = {}

    for i,num in enumerate(nums):
        if num in map3:
            return True
        else:
            map3[num] = i
            if len(map3) > k:
                del map3[nums[i-k]]
        print(f"Current Map: {map3}")
    return False

print(contains_nearby_duplicate(nums, k))
```

### 3] Valid Anagram

Use the Hashmap "get(s[i],0)" function

```python
s = "cat"
t = "tac"

def is_anagram(s,t):
    if len(s) != len(t):
        return False
    
    map4 = {}
    for i in range(len(s)):
        map4[s[i]] = map4.get(s[i],0) + 1
        map4[t[i]] = map4.get(t[i],0) - 1
        print(f"Current Map: {map4}")
    if all(value == 0 for value in map4.values()):
        return True
    else:
        return False

print(is_anagram(s,t))
```

### 5] Group Anagram

```python
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]

def group_anagrams(strs):
    map5 = {}

    for s in strs:
        sorted_s = ''.join(sorted(s))
        if sorted_s in map5:
            map5[sorted_s].append(s)
        else:
            map5[sorted_s] = [s]
        print(f"Current Map: {map5}")
    return list(map5.values())

print(group_anagrams(strs))
```

### 6] Product of Array except SELF

```python
# Return the product of all the elements of nums except nums[i]
nums = [1,2,3,4,5]

def product(nums):
    prefix,suffix = 1,1

    map6 = {}
    result = [1] * len(nums)

    for i in range(len(nums)):
        result[i] = prefix
        prefix *= nums[i]

    for i in range(len(nums)-1,-1,-1):
        result[i] *= suffix
        suffix *= nums[i]

    return result

print(product(nums))
```

### 7] Top K Frequent Elements

Using Priority Queue - ".put()", ".get()", ".empty()"

```python
from queue import PriorityQueue

nums = [1,3,4,3,4,2,3,4,2,5,4,5,5]
k = 3

def top_k_frequent(nums, k):
    if k == len(nums):
        return nums

    map7 = {}
    pq = PriorityQueue()
    
    for num in nums:
        map7[num] = map7.get(num,0) + 1
        print(f"Current Map: {map7}")

    for num, freq in map7.items():
        pq.put((freq, num))
        print(f"Current Priority Queue: {pq.queue}")

        if pq.qsize() > k:
            pq.get()
            print(f"Current Priority Queue after removal: {pq.queue}")

    result = []
    while not pq.empty():
        result.append(pq.get()[1])
        print(f"Current Result: {result}")

    return result

print(top_k_frequent(nums, k))    
```
